```python
# -----------------------------------------
#  SIMULADOR INTERACTIVO DE TENIS DE MESA
# -----------------------------------------

import sys
sys.path.insert(0, "../src")

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

from IPython.display import display, clear_output
from table_tennis_sim.simulation import simulate


# -----------------------------------------
#  CONTROLES
# -----------------------------------------

vx_slider = widgets.FloatSlider(
    value=3000,
    min=500,
    max=5000,
    step=100,
    description="Vx:",
    continuous_update=False
)

vy_slider = widgets.FloatSlider(
    value=0,
    min=-2000,
    max=2000,
    step=100,
    description="Vy:",
    continuous_update=False
)

vz_slider = widgets.FloatSlider(
    value=-3000,
    min=-5000,
    max=2000,
    step=100,
    description="Vz:",
    continuous_update=False
)

wx_slider = widgets.FloatSlider(
    value=0,
    min=-50,
    max=50,
    step=1,
    description="ωx:",
    continuous_update=False
)

wy_slider = widgets.FloatSlider(
    value=0,
    min=-50,
    max=50,
    step=1,
    description="ωy:",
    continuous_update=False
)

wz_slider = widgets.FloatSlider(
    value=10,
    min=-50,
    max=50,
    step=1,
    description="ωz:",
    continuous_update=False
)


# -----------------------------------------
#  BOTÓN
# -----------------------------------------

simulate_button = widgets.Button(
    description="▶ Simular",
    button_style="success"
)

output = widgets.Output()


# -----------------------------------------
#  FUNCIÓN DE SIMULACIÓN
# -----------------------------------------

def run_simulation(_):

    with output:
        clear_output(wait=True)

        # ----------------------------------------------------
        # Condiciones iniciales
        # ----------------------------------------------------

        p0 = np.array([
            1000.0,
            762.5,
            1065.0
        ])

        v0 = np.array([
            vx_slider.value,
            vy_slider.value,
            vz_slider.value
        ])

        w0 = np.array([
            wx_slider.value,
            wy_slider.value,
            wz_slider.value
        ])

        # ----------------------------------------------------
        # Ejecutar simulación
        # ----------------------------------------------------

        t, pos, vel, acc, omega, alpha = simulate(
            p0,
            v0,
            w0
        )

        # ----------------------------------------------------
        # Información
        # ----------------------------------------------------

        print(" Simulación ejecutada correctamente")
        print()
        print(f"Velocidad inicial: {v0}")
        print(f"Rotación inicial: {w0}")
        print(f"Tiempo simulado: {t[-1]:.3f} s")


        # -----------------------------------------
        #  GRÁFICA 1 — TRAYECTORIA 3D
        # -----------------------------------------

        fig = plt.figure(figsize=(10, 6))
        ax = fig.add_subplot(111, projection="3d")

        ax.plot(
            pos[0],
            pos[1],
            pos[2],
            linewidth=2,
            label="Trayectoria de la pelota"
        )

        ax.set_xlabel("X [mm]")
        ax.set_ylabel("Y [mm]")
        ax.set_zlabel("Z [mm]")
        ax.set_title("Trayectoria 3D de la pelota")
        ax.legend()

        plt.show()


        # -----------------------------------------
        #  GRÁFICA 2 — POSICIÓN
        # -----------------------------------------

        plt.figure(figsize=(10, 5))

        plt.plot(t, pos[0], label="X")
        plt.plot(t, pos[1], label="Y")
        plt.plot(t, pos[2], label="Z")

        plt.xlabel("Tiempo [s]")
        plt.ylabel("Posición [mm]")
        plt.title("Posición de la pelota")
        plt.legend()
        plt.grid(True)

        plt.show()


        # -----------------------------------------
        #  GRÁFICA 3 — VELOCIDAD
        # -----------------------------------------

        plt.figure(figsize=(10, 5))

        plt.plot(t, vel[0], label="Vx")
        plt.plot(t, vel[1], label="Vy")
        plt.plot(t, vel[2], label="Vz")

        plt.xlabel("Tiempo [s]")
        plt.ylabel("Velocidad [mm/s]")
        plt.title("Velocidad de la pelota")
        plt.legend()
        plt.grid(True)

        plt.show()


# -----------------------------------------
#  CONECTAR BOTÓN
# -----------------------------------------

simulate_button.on_click(run_simulation)


# -----------------------------------------
#  MOSTRAR INTERFAZ
# -----------------------------------------

display(
    widgets.HTML(
        "<h2> Simulador interactivo de tenis de mesa</h2>"
    ),
    widgets.HTML(
        "<b>Velocidad inicial</b>"
    ),
    vx_slider,
    vy_slider,
    vz_slider,
    widgets.HTML(
        "<b>Rotación inicial</b>"
    ),
    wx_slider,
    wy_slider,
    wz_slider,
    simulate_button,
    output
)
```

#  Simulador interactivo de tenis de mesa

Este simulador permite analizar el movimiento de una pelota de tenis de mesa
a partir de diferentes condiciones iniciales de velocidad y rotación.
##  Parámetros de entrada

Los controles permiten modificar:

- **Vx:** velocidad inicial en el eje X.
- **Vy:** velocidad inicial en el eje Y.
- **Vz:** velocidad inicial en el eje Z.
- **ωx:** velocidad angular inicial alrededor del eje X.
- **ωy:** velocidad angular inicial alrededor del eje Y.
- **ωz:** velocidad angular inicial alrededor del eje Z.

Al presionar el botón ** Simular**, el modelo utiliza estos valores para
calcular el movimiento de la pelota.

##  Resultados de la simulación

El simulador genera tres gráficas:

1. **Trayectoria 3D:** representa el recorrido espacial de la pelota.
2. **Posición vs. tiempo:** muestra la evolución de las coordenadas X, Y y Z.
3. **Velocidad vs. tiempo:** muestra la evolución de las componentes Vx, Vy y Vz.

Estas gráficas permiten observar cómo las condiciones iniciales afectan el
movimiento de la pelota.

##  Análisis

Al modificar la velocidad inicial y la rotación de la pelota se pueden
comparar diferentes trayectorias y estudiar el efecto que tienen estos
parámetros sobre el comportamiento del sistema.
